```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv(url, names=columns)

df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
glucose_mean = df["Glucose"].mean()

print("Sample Mean of Glucose:", glucose_mean)
```

```python
diabetes_count = df["Outcome"].sum()
total_patients = len(df)

sample_proportion = diabetes_count / total_patients

print("Number of diabetic patients:", diabetes_count)
print("Total patients:", total_patients)
print("Sample proportion:", sample_proportion)
print("Sample proportion (%):", sample_proportion * 100)
```

```python
sample_mean = df["Glucose"].mean()
sample_std = df["Glucose"].std()
n = len(df)

confidence_level = 0.95
alpha = 1 - confidence_level

standard_error = sample_std / np.sqrt(n)

t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

margin_of_error = t_critical * standard_error

lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print("Sample Mean:", sample_mean)
print("Standard Deviation:", sample_std)
print("Standard Error:", standard_error)
print("95% Confidence Interval:")
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
reference_value = 120

t_statistic, p_value = stats.ttest_1samp(
    df["Glucose"],
    reference_value
)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Glucose"], kde=True)

plt.axvline(120, linestyle="--", label="Reference Mean = 120")

plt.title("Distribution of Glucose Levels")
plt.xlabel("Glucose Level")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```
